# Formulacion de un emprendimiento tecnologico

### Instrucciones

En parejas desarrolla la siguiente guia, deben copiar en un solo cuaderno, pero la competencia, ambito y objetivo si debe aparecer en ambos

1. Identificamos la problemática

Seleccionen una problemática o necesidad de su entorno que pueda ser atendida mediante una solución tecnológica. Describan:

* ¿Cuál es el problema?
* ¿A quiénes afecta?
* ¿Por qué consideran que es importante buscar una solución?

2. Proponemos la solución tecnológica

Planteen una idea de producto, servicio, aplicación, plataforma u otra solución tecnológica que pueda responder a la problemática.

* ¿Qué solución proponen?
* ¿Cómo funcionaría de manera general?
* ¿Qué beneficio ofrecería a las personas afectadas?

3. Definimos nuestro emprendimiento

Completen los elementos básicos de la propuesta:
* Nombre del emprendimiento.
* Producto o servicio.
* ¿Quiénes serían sus usuarios o clientes?
* ¿Qué necesidad resolvería?
* ¿Qué tendría de diferente o útil frente a otras alternativas?

4. Identificamos los recursos y la información financiera

Piensa en lo que necesitarías para poner en funcionamiento tu emprendimiento.

Recursos necesarios:

* Recursos tecnológicos: ______________________________
* Recursos humanos: ___________________________________
* Otros recursos: _____________________________________

>
> Objetivo: Construir una primera propuesta de emprendimiento tecnológico a partir de una problemática real o cercana.
